```python
import pandas as pd
from functions import wormstats

all_data = []

for x in range(1, 13):
    csv_file_path = f'companyDrug/companyDrug{x}.csv'  # Use f-string for cleaner code
    y = wormstats(csv_file_path)  # Process the file
    y['drugged'] = 1            # Add the 'drugged' column
    y['worm_id'] = x             # Add the 'worm_id' column
    all_data.append(y)           # Append the DataFrame to the list

for x in range(1, 13):
    csv_file_path = f'controlTerbinafin/controlTerbinafin{x}.csv'  # Use f-string for cleaner code
    y = wormstats(csv_file_path)  # Process the file
    y['drugged'] = 0             # Add the 'drugged' column
    y['worm_id'] = x+12             # Add the 'worm_id' column
    all_data.append(y)           # Append the DataFrame to the list

for x in range(1, 13):
    csv_file_path = f'Terbinafin/Terbinafin{x}.csv'  # Use f-string for cleaner code
    y = wormstats(csv_file_path)  # Process the file
    y['drugged'] = 2            # Add the 'drugged' column
    y['worm_id'] = x+24             # Add the 'worm_id' column
    all_data.append(y)           # Append the DataFrame to the list

for x in range(1, 13):
    csv_file_path = f'control/control{x}.csv'  # Use f-string for cleaner code
    y = wormstats(csv_file_path)  # Process the file
    y['drugged'] = 0             # Add the 'drugged' column
    y['worm_id'] = x+36            # Add the 'worm_id' column
    all_data.append(y)           # Append the DataFrame to the list

merged_df = pd.concat(all_data, axis=0, ignore_index=True)
merged_df
```

```python
merged_df.to_csv('mergedworms.csv')
```
